```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

datasets = [
    ("Config8",  "250# bedding-in", [3348 - 3316], 250),
    ("Config8",  "100# test",       [3316 - 3308], 100),

    ("Config9",  "250# bedding-in", [3004.0 - 2949.0], 250),
    ("Config9",  "100# test",       [2949.0 - 2935.0], 100),

    ("Config10", "350# bedding-in", [7700.2 - 7566.2], 350),
    ("Config10", "100# test",       [7566.2 - 7558.6], 100),
]

# Same Y-axis limit for all plots
global_max = max(max(wear_values) for _, _, wear_values, _ in datasets)
ymax = global_max * 1.15

for config, test, wear_values, n_brakings in datasets:

    total_wear = sum(wear_values)
    wear_per_braking = total_wear / n_brakings

    plt.figure(figsize=(6, 5))

    x = [0]

    bars = plt.bar(
        x,
        wear_values,
        width=0.15,
        color='steelblue',
        edgecolor='black'
    )

    plt.xticks(x, ['1 Bg'])
    plt.xlim(-0.5, 0.5)

    plt.ylim(0, ymax)

    plt.xlabel('Block', fontsize=12, fontweight='bold')
    plt.ylabel('Block wear [g]', fontsize=12, fontweight='bold')
    plt.title(f'{config} - {test}', fontsize=14, fontweight='bold')

    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Value above bar
    for bar in bars:
        height = bar.get_height()

        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + ymax * 0.01,
            f'{height:.1f}',
            ha='center',
            va='bottom',
            fontsize=12,
            fontweight='bold'
        )

    # Info box
    plt.text(
        0.65,
        0.85,
        f'Total wear = {total_wear:.1f} g\n'
        f'N° braking = {n_brakings}\n'
        f'Wear/braking = {wear_per_braking:.3f} g',
        transform=plt.gca().transAxes,
        fontsize=11,
        fontweight='bold',
        bbox=dict(
            facecolor='white',
            alpha=0.85,
            edgecolor='gray',
            boxstyle='round,pad=0.4'
        )
    )

    plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

Config = "Config11"

# Data
wear_bedding = [2979 - 2914, 2970 - 2919]  # Up, Down
wear_test    = [2914 - 2911, 2919 - 2917]  # Up, Down

positions = ['Up', 'Down']

datasets = {
    "1000# bedding-in": (wear_bedding, 1000),
    "100# test": (wear_test, 100)
}

for test_name, (wear_values, n_brakings) in datasets.items():

    total_wear = sum(wear_values)
    wear_per_braking = total_wear / n_brakings

    plt.figure(figsize=(6, 6))
    plt.style.use('seaborn-v0_8-whitegrid')

    bars = plt.bar(
        positions,
        wear_values,
        color='steelblue',
        edgecolor='black',
        width=0.6
    )

    plt.xlabel('Block 2 Bgu position', fontsize=12, fontweight='bold')
    plt.ylabel('Block wear [g]', fontsize=12, fontweight='bold')
    plt.title(f'{Config} - {test_name}', fontsize=14, fontweight='bold')
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Labels on bars
    for bar in bars:
        height = bar.get_height()

        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + max(wear_values) * 0.01,
            f'{height:.1f}',
            ha='center',
            va='bottom',
            fontsize=12,
            fontweight='bold'
        )

    # Information box
    plt.text(
        0.65,
        0.85,
        f'Total wear = {total_wear:.1f} g\n'
        f'N° brakings = {n_brakings}\n'
        f'Wear/braking = {wear_per_braking:.3f} g',
        transform=plt.gca().transAxes,
        fontsize=11,
        fontweight='bold',
        bbox=dict(
            facecolor='white',
            alpha=0.8,
            edgecolor='gray',
            boxstyle='round,pad=0.5'
        )
    )

    plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

# (vehicle, config, total wear in 100# test [g], n_brakings)
all_configs = [
    ("X61",           "Config1",      19.2, 100),
    ("X61",           "Config1_rep2", 20.2, 100),
    ("X61",           "Config2",      81.8, 100),
    ("AGC",           "Config3",      39.8, 100),
    ("AGC",           "Config4",      30.2, 100),
    ("AGC",           "Config5",      38.6, 100),
    ("B20",           "Config6",      27.6, 100),
    ("B20",           "Config7",      21.0, 100),
    ("MF19",          "Config8",       8.0, 100),
    ("MF19",          "Config9",      14.0, 100),
    ("MF19",          "Config10",      7.6, 100),
    ("Traxx Shunter", "Config11",      5.0, 100),
]

vehicles = [v for v, *_ in all_configs]
configs = [c for _, c, *_ in all_configs]
wear_per_braking = [w / n for _, _, w, n in all_configs]

unique_vehicles = list(dict.fromkeys(vehicles))
cmap = plt.get_cmap('tab10')
vehicle_colors = {v: cmap(i) for i, v in enumerate(unique_vehicles)}
bar_colors = [vehicle_colors[v] for v in vehicles]

plt.figure(figsize=(12, 6))

bars = plt.bar(configs, wear_per_braking, color=bar_colors, edgecolor='black')

plt.xlabel('Config', fontsize=12, fontweight='bold')
plt.ylabel('Wear/braking [g]', fontsize=12, fontweight='bold')
plt.title('Wear per braking - 100# test (all configs)', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar, value in zip(bars, wear_per_braking):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + max(wear_per_braking) * 0.01,
        f'{value:.3f}',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

legend_handles = [plt.Rectangle((0, 0), 1, 1, color=vehicle_colors[v]) for v in unique_vehicles]
plt.legend(legend_handles, unique_vehicles, title='Project', loc='upper right')

plt.tight_layout()
plt.show()
```
