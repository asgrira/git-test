```python
# Liste pour stocker les DataFrames
dfs = []

# Parcourir tous les fichiers .dth
for filename in os.listdir(file):
    if filename.endswith(".dth"):
        file_path = os.path.join(file, filename)

        # Lire le fichier avec l'encodage approprié
        try:
            df = pd.read_csv(file_path, sep=";", encoding="latin-1")
        except UnicodeDecodeError:
            df = pd.read_csv(file_path, sep=";", encoding="ISO-8859-1")

        # Garder uniquement les colonnes T1 à T3 (noms réels, ex: "T1(°C)"), hors TEMPS(sec)
        cols_needed = [col for col in df.columns if col.startswith("T") and col != "TEMPS(sec)"]
        cols_needed = cols_needed[:3]  # T1 à T3

        # Extraire ces colonnes ainsi que la colonne TEMPS
        df_selected = df[cols_needed + ["TEMPS(sec)"]].copy()

        # Ajouter une colonne pour identifier le fichier source
        df_selected["source_file"] = filename

        # Ajouter au tableau global
        dfs.append(df_selected)

# Concaténer tous les DataFrames
final_df = pd.concat(dfs, ignore_index=True)

# Exporter en CSV
output_path = os.path.join(file, "base_T1_T3.csv")
final_df.to_csv(output_path, index=False)

print(f"Base de données créée : {output_path}")
```

#3

```python
# Liste pour stocker les DataFrames
dfs = []
last_time = 0.0
d = 9.5  # ajuster le décalage temps entre fichiers température disque et ELPI

# Parcourir tous les fichiers .dth
for filename in os.listdir(file):
    if filename.endswith(".dth"):
        file_path = os.path.join(file, filename)

        # Lire le fichier avec encodage approprié
        try:
            df = pd.read_csv(file_path, sep=";", encoding="latin-1")
        except UnicodeDecodeError:
            df = pd.read_csv(file_path, sep=";", encoding="ISO-8859-1")

        # Garder uniquement T1, T2 et T3 (noms réels, ex: "T1(°C)"), hors TEMPS(sec)
        cols_needed = [col for col in df.columns if col.startswith("T") and col != "TEMPS(sec)"]
        cols_needed = cols_needed[:3]

        # Ajouter la colonne TEMPS ajustée
        df["TEMPS(sec)"] = df["TEMPS(sec)"] + last_time
        last_time = df["TEMPS(sec)"].iloc[-1] + d + 24.6 + 74 + 13.9

        # Sélectionner T1-T3 + TEMPS
        df_selected = df[cols_needed + ["TEMPS(sec)"]].copy()

        # Ajouter une colonne pour identifier le fichier source
        df_selected["source_file"] = filename

        # Ajouter au tableau global
        dfs.append(df_selected)

# Concaténer tous les DataFrames
final_df = pd.concat(dfs, ignore_index=True)

# Exporter en CSV
output_path_inc = os.path.join(file, "base_T1_T3_inc.csv")
final_df.to_csv(output_path_inc, index=False)

print(f"Base de données créée : {output_path_inc}")
```
