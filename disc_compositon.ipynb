Disc FT 20 composition 

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm

# Composition chimique (valeurs moyennes)
composition = {
    "C": (3.20 + 3.60) / 2,
    "Si": (2.00 + 2.20) / 2,
    "Mn": (0.60 + 0.80) / 2,
    "Mo": 0.3,
    "Cr": (0.30 + 0.45) / 2,
    "Ni": 0.1,
    "S": (0.08 + 0.10) / 2,
    "P": 0.1,
    "Cu": (0.40 + 1.00) / 2,
    "Sn": 0.1
}

# ---- AJOUT DU FER ----
total_without_fe = sum(composition.values())
Fe_value = 100 - total_without_fe
composition["Fe"] = Fe_value

# ---- TRI du plus élevé au plus faible ----
composition_sorted = dict(
    sorted(composition.items(), key=lambda x: x[1], reverse=True)
)

labels = list(composition_sorted.keys())
values = list(composition_sorted.values())

# ---- Couleurs uniques sans répétition ----
# Utilisation d’un colormap pour générer N couleurs distinctes
cmap = cm.get_cmap('tab20', len(labels))  # 20 couleurs distinctes
colors = [cmap(i) for i in range(len(labels))]

# Remplacer couleurs pour C et Fe :
for i, elem in enumerate(labels):
    if elem == "C":
        colors[i] = "#1f77b4"   # bleu foncé
    if elem == "Fe":
        colors[i] = "#374049"   # bleu clair

# --------- GRAPHIQUE ---------
fig, ax = plt.subplots(figsize=(10, 8))

wedges, _ = ax.pie(
    values,
    startangle=90,
    wedgeprops={'width': 0.35},
    colors=colors
)

# --------- LÉGENDE ---------
labels_with_values = [
    f"{elem} : {value:.2f}%" for elem, value in composition_sorted.items()
]

plt.legend(
    wedges,
    labels_with_values,
    title="Chemical elements",
    loc="center left",
    bbox_to_anchor=(1, 0.5),
    fontsize=12
)

plt.title("Cast iron FT20 IBRE", fontsize=18, fontweight='bold')
plt.axis('equal')
plt.tight_layout()

plt.show()
```
