```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# =====================================================
# FILE
# =====================================================

file_path = r"C:\Users\Grira\OneDrive - Alstom\Documents\Paddle\Campagnes3_4\Banc2_COM\COM_Camp 3&4_PadDLE.xlsx"

sheet_name = "RIB_PS"

df = pd.read_excel(
    file_path,
    sheet_name=sheet_name,
    header=None
)

# =====================================================
# CONFIGURATIONS
# =====================================================

configs = {
    "Config8": 87,
    "Config9": 92,
    "Config10": 97,
    "Config11": 102
}

# =====================================================
# TESTS
# =====================================================

tests = {
    "RIB-PS": {
        "start_row": 18,
        "n": 100
    },

    "Bedding": {
        "start_row": 325,
        "n": {
            "Config8": 250,
            "Config9": 250,
            "Config10": 350,
            "Config11": 1000
        }
    }
}

# =====================================================
# FUNCTION
# =====================================================

def plot_test(config_name, test_name):

    um_col = configs[config_name]

    start_row = tests[test_name]["start_row"]

    if test_name == "Bedding":
        n = tests[test_name]["n"][config_name]
    else:
        n = tests[test_name]["n"]

    # Extraction des données
    mu = pd.to_numeric(
        df.iloc[start_row:start_row+n, um_col],
        errors='coerce'
    )

    mu = mu.dropna()

    cycles = np.arange(1, len(mu) + 1)

    mean_mu = np.mean(mu)
    std_mu = np.std(mu, ddof=1)

    print("\nPremiers 5 points :")
    for i, value in enumerate(mu.head(), start=1):
        print(f"Braking {i} : µ = {value:.3f}")

    print(f"\n{config_name} - {test_name}")
    print(f"Nombre de points : {len(mu)}")
    print(f"Mean : {mean_mu:.3f}")
    print(f"Std  : {std_mu:.3f}")

    # Plot
    plt.figure(figsize=(10, 6))

    plt.plot(
        cycles,
        mu,
        marker='o',
        markersize=3,
        linewidth=1.2,
        color='blue',
        label='µ'
    )

    plt.axhline(
        mean_mu,
        color='red',
        linestyle='--',
        linewidth=2,
        label=f'Mean µ ± σ = {mean_mu:.3f} ± {std_mu:.3f}'
    )

    plt.title(f"{config_name} - {test_name}")

    plt.xlabel("Braking number")
    plt.ylabel("Friction coefficient (µ)")

    plt.grid(True, linestyle='--', alpha=0.5)
    plt.ylim(0, 0.6)
    plt.legend()

    plt.xlim(0, len(mu) + 10)

    plt.tight_layout()

    plt.show()


# =====================================================
# PLOT ALL
# =====================================================

for config_name in configs.keys():

    plot_test(config_name, "Bedding")

    plot_test(config_name, "RIB-PS")
```
